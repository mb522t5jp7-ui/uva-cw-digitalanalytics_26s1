Below you have a list of visitors , and in particular their referral. Imagine that this list of referrals to the website of a news organization is provided by Google Analytics. Create a function that categorizes the visitors according to some interesting set of categories.

```python
visitors = ['campaign utx=9902',
             'instagram app',
             'facebook app',
             'campaign utx=1389',
             'facebook app',
             'google search',
             'newsletter',
             'facebook app',
             'newsletter',
             'campaign utx=9902',
             'instagram app',
             'campaign utx=1389',
             'newsletter',
             'facebook.com',
             'facebook app',
             'newsletter',
             'google search',
             'campaign utx=9902',
             'campaign utx=9902',
             'instagram app',
             'organic',
             'instagram app',
             'news campaign 9902',
             'instagram app',
             'facebook app',
             'facebook.com',
             'campaign utx=9902',
             'campaign utx=9902',
             'campaign utx=1389',
             'campaign utx=1389',
             'facebook.com',
             'news campaign 9902',
             'newsletter',
             'instagram app',
             'instagram app',
             'campaign utx=1389',
             'direct',
             'facebook.com',
             'newsletter',
             'direct',
             'campaign utx=1389',
             'direct',
             'organic',
             'facebook.com',
             'facebook.com',
             'facebook app',
             'campaign utx=9902',
             'google search',
             'campaign utx=9902',
             'campaign utx=1389']
```

```python
# Check visitors
def check_visitors(x):
    if 'facebook' in x:
        return 'social media'
    if 'campaign' in x:
        return 'advertisement'
    if x in ['direct', 'organic', 'google search']:
        return 'referrals'
    if 'news' in x:
        return 'newsletter'
    if 'instagram' in x:
        return 'social media'
    return 'not categorized'
```

```python
# Catagorizes visitors
def catagorize_visitors(visitors):
    list_catagory = []
    for visitor in visitors:
        # print(check_visitors(visitor))
        if 'facebook' in visitor:
            list_catagory.append('social media')
        if 'campaign' in visitor:
            list_catagory.append('advertisement')
        if visitor in ['direct', 'organic', 'google search']:
            list_catagory.append('referrals')
        if 'news' in visitor:
            list_catagory.append('newsletter')
        if 'instagram' in visitor:
            list_catagory.append('newsletter')
    return list_catagory
```

```python
catagorized_list = catagorize_visitors(visitors)
```

### Code part-2 Analysis — From the Weekly Challenges

Write a code that further analyzes the categorized data.

1. Count how many times each source appears.
2. Calculate the percentage of customers coming from each source.
3. Identify the most popular source.

```python
# 1-Create a dictionary to organize the count
uniq_list = set(catagorized_list) 
dict_catagory_count = {}
for item in uniq_list:
    dict_catagory_count[item] = catagorized_list.count(item)
dict_catagory_count
```

```python
# percentage of customers coming from each source
for k,v in dict_catagory_count.items():
    print(f'Percentage of {k} is {100*(v/len(catagorized_list))}')
```

```python
# the most popular source
max(dict_catagory_count, key=dict_catagory_count.get)
```
